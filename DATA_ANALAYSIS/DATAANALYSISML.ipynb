DATA ANALYSIS WITH PANDAS


```python
import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Cathy"],
    "Marks": [80, 65, 90]
}

df = pd.DataFrame(data)

print(df)
```

DATAFRAME


```python
import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Cathy", "David"],
    "Marks": [80, 65, 90, 75],
    "City": ["Chennai", "Madurai", "Salem", "Chennai"]
}

df = pd.DataFrame(data)

print(df)

print(df["Name"])

print(df[["Name", "Marks"]])
```

COMBINING

```python
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Arun", "Bala", "Cathy"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 70, 90]
})

result = pd.merge(students, marks, on="ID")

print(result)
```

INDEXING

```python
print(df.loc[0])

print(df.loc[0, "Marks"])

print(df.iloc[0])

print(df.iloc[0, 1])
```

GROUPING


```python
data = {
    "Department": ["CSE", "CSE", "ECE", "ECE"],
    "Marks": [80, 70, 90, 60]
}

df = pd.DataFrame(data)

print(df.groupby("Department")["Marks"].mean())

print(df.groupby("Department")["Marks"].sum())

print(df.groupby("Department")["Marks"].max())

print(df.groupby("Department")["Marks"].min())

print(df.groupby("Department")["Marks"].count())
```

FILTERING

```python
import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Cathy", "David"],
    "Marks": [80, 65, 90, 75],
    "City": ["Chennai", "Madurai", "Salem", "Chennai"]
}

df = pd.DataFrame(data)

print(df[df["Marks"] > 70])

print(df[df["Marks"] >= 75])

print(df[df["City"] == "Chennai"])

print(df[(df["Marks"] > 70) & (df["City"] == "Chennai")])

print(df[(df["Marks"] > 80) | (df["City"] == "Chennai")])
```

SORTING

```python
df.sort_values("Marks")

df.sort_values("Marks", ascending=False)

df.sort_values("Name")

df.sort_values(["City", "Marks"])
```

PLOTTING

```python
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Cathy", "David"],
    "Marks": [80, 65, 90, 75]
})

# Bar Chart
df.plot(x="Name", y="Marks", kind="bar")
plt.show()

# Line Chart
df.plot(x="Name", y="Marks", kind="line")
plt.show()

# Pie Chart
df.plot(
    y="Marks",
    kind="pie",
    labels=df["Name"],
    autopct="%1.1f%%"
)
plt.show()

# Histogram
df["Marks"].plot(kind="hist")
plt.show()
```

DATA ANALYSIS ALL IN ONE.

```python
import pandas as pd
import matplotlib.pyplot as plt

students = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Name": ["Arun", "Bala", "Cathy", "David", "Eva"],
    "Department": ["CSE", "ECE", "CSE", "ECE", "CSE"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Marks": [85, 65, 90, 70, 78]
})

df = pd.merge(students, marks, on="ID")

print(df)

print(df.iloc[0])

print(df[["Name", "Marks"]])

print(df[df["Marks"] > 75])

print(df.groupby("Department")["Marks"].mean())

sorted_df = df.sort_values("Marks", ascending=False)

print(sorted_df)

df.to_csv("student_analysis.csv", index=False)

df.plot(x="Name", y="Marks", kind="bar")
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```
