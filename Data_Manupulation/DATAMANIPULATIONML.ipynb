RANDOM NUMBERS


```python
import numpy as np

print("----- RANDOM NUMBERS -----")

# Random decimal
print("Random Decimal:")
print(np.random.random())

# Five random decimals
print("\nFive Random Decimals:")
print(np.random.random(5))

# Random integer
print("\nRandom Integer (1-9):")
print(np.random.randint(1,10))

# Dice
print("\nDice Roll:")
print(np.random.randint(1,7))

# Ten random integers
print("\n10 Random Numbers:")
print(np.random.randint(1,101,10))

# Random choice
students = np.array(["Arun","Bala","John","Priya"])
print("\nRandom Student:")
print(np.random.choice(students))

# Two random students
print("\nTwo Random Students:")
print(np.random.choice(students,2))

# Shuffle
numbers = np.array([1,2,3,4,5])
np.random.shuffle(numbers)
print("\nShuffled Array:")
print(numbers)

# Random seed
np.random.seed(42)
print("\nRandom Seed:")
print(np.random.randint(1,100,5))
```

DATA MANIPULATION AND INDEXING

```python
import numpy as np

print("----- DATA MANIPULATION & INDEXING -----")

marks = np.array([70,80,90,60,75])

print("Original Array:", marks)

print("\nPositive Indexing")
print(marks[0])
print(marks[3])

print("\nNegative Indexing")
print(marks[-1])
print(marks[-2])

print("\nSlicing")
print(marks[1:4])

a = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print("\n2D Array")
print(a)

print("\nElement [0,0]:", a[0,0])
print("Element [1,2]:", a[1,2])
print("Element [2,1]:", a[2,1])

print("\nSecond Row")
print(a[1])

print("\nSecond Column")
print(a[:,1])
```

FILTERING


```python
import numpy as np

print("----- FILTERING -----")

marks = np.array([45,78,90,55,82,60,70])

print("Original:", marks)

print("\nGreater than 60")
print(marks[marks>60])

print("\nLess than 60")
print(marks[marks<60])

print("\nGreater than or Equal to 60")
print(marks[marks>=60])

print("\nEqual to 60")
print(marks[marks==60])

print("\nNot Equal to 60")
print(marks[marks!=60])

print("\nBetween 60 and 80")
print(marks[(marks>=60)&(marks<=80)])

print("\nLess than 50 OR Greater than 80")
print(marks[(marks<50)|(marks>80)])

numbers=np.array([1,2,3,4,5,6,7,8])

print("\nEven Numbers")
print(numbers[numbers%2==0])

print("\nOdd Numbers")
print(numbers[numbers%2!=0])
```

STATISTICS


```python
import numpy as np

print("----- STATISTICS -----")

data = np.array([10,20,30,40,50])

print("Data:", data)

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Range:", np.max(data)-np.min(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))

print("25th Percentile:", np.percentile(data,25))
print("50th Percentile:", np.percentile(data,50))
print("75th Percentile:", np.percentile(data,75))
```

AGGREGATION


```python
import numpy as np

print("----- AGGREGATION -----")

sales = np.array([100,200,150,300,250])

print("Sales:", sales)

print("Total:", np.sum(sales))
print("Average:", np.mean(sales))
print("Minimum:", np.min(sales))
print("Maximum:", np.max(sales))
print("Number of Records:", sales.size)

data = np.array([
    [10,20,30],
    [40,50,60]
])

print("\n2D Array")
print(data)

print("\nTotal Sum")
print(np.sum(data))

print("\nColumn-wise Sum")
print(np.sum(data,axis=0))

print("\nRow-wise Sum")
print(np.sum(data,axis=1))

print("\nColumn-wise Mean")
print(np.mean(data,axis=0))
```

SAVING DATA

```python
import numpy as np

print("----- SAVING DATA -----")

marks = np.array([70,80,90,60,75])

# Save as NumPy file
np.save("marks.npy", marks)
print("marks.npy saved")

# Load NumPy file
loaded = np.load("marks.npy")
print("Loaded:", loaded)

# Save as Text
np.savetxt("marks.txt", marks)
print("marks.txt saved")

# Load Text
loaded_text = np.loadtxt("marks.txt")
print("Loaded Text:", loaded_text)

# Save CSV
data = np.array([
    [101,75],
    [102,85],
    [103,90]
])

np.savetxt("data.csv", data, delimiter=",", fmt="%d")
print("data.csv saved")

# Load CSV
loaded_csv = np.loadtxt("data.csv", delimiter=",")
print("Loaded CSV:")
print(loaded_csv)
```
