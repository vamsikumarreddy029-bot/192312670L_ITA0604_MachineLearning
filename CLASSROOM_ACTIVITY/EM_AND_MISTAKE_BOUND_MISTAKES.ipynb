```python
import numpy as np
from sklearn.mixture import GaussianMixture

# Student marks data
marks = np.array([85, 78, 82, 58, 62, 55, 79, 61]).reshape(-1, 1)

# EM model: 2 hidden groups
model = GaussianMixture(n_components=2, random_state=0)

# Train model
model.fit(marks)

# Predict group
groups = model.predict(marks)

print("Marks:", marks.ravel())
print("Predicted Groups:", groups)

print("\nGroup Means:")
print(model.means_)

print("\nGroup Probabilities:")
print(model.weights_)

```

## Mistake Bound Model (Panda vs Tiger Example)

```python
# ==========================================
# SIMPLE MISTAKE BOUND MODEL
# Panda vs Tiger Example
# ==========================================

# Animal size score
# Negative values = small animal
# Positive values = large animal

X = [-3, -2, -1, 1, 2, 3]

# Actual answer
# -1 = Panda
#  1 = Tiger
y = [-1, -1, -1, 1, 1, 1]

# Initially model knows nothing
weight = 0

mistakes = 0

print("PANDA = -1")
print("TIGER = +1")
print("--------------------------")

for i in range(len(X)):

    # Model prediction
    score = weight * X[i]

    if score >= 0:
        prediction = 1      # TIGER
    else:
        prediction = -1     # PANDA

    # Convert numbers to names
    actual_name = "TIGER" if y[i] == 1 else "PANDA"
    predicted_name = "TIGER" if prediction == 1 else "PANDA"

    print("\nAnimal", i + 1)
    print("Actual     :", actual_name)
    print("Prediction :", predicted_name)

    # Check mistake
    if prediction != y[i]:

        mistakes += 1

        print(" MISTAKE")

        # Learn from mistake
        weight = weight + y[i] * X[i]

        print("Model learned! New weight =", weight)

    else:
        print(" CORRECT")

print("\n========================")
print("TOTAL MISTAKES =", mistakes)
print("========================")
```

```python
2. # FINITE AND INFINITE HYPOTHESIS SPACE

# =====================================================
# FINITE AND INFINITE HYPOTHESIS SPACE
# SIMPLE SHIRT EXAMPLE
# =====================================================

print("========== FINITE HYPOTHESIS SPACE ==========")

# Ready-made shirts
sizes = ["S", "M", "L", "XL"]

print("Ready-made shirt sizes:", sizes)

print("Total choices:", len(sizes))

customer_choice = "M"

print("Customer selected:", customer_choice)

if customer_choice in sizes:
    print("Selected shirt is available.")

print("\nFINITE means:")
print("Only fixed choices are available.")
print("Hypothesis Space =", sizes)

print("\n\n========== INFINITE HYPOTHESIS SPACE ==========")

print("Now customer wants a CUSTOM-MADE shirt.")

customer_length = 72.35

print("Customer shirt length:", customer_length, "cm")

print("\nPossible custom measurements can be:")

examples = [
    72,
    72.1,
    72.12,
    72.123,
    72.1234
]

for measurement in examples:
    print(measurement, "cm")

print("\nAnd we can continue:")
print("72.12345")
print("72.123456")
print("72.1234567")
print("...")

print("\nINFINITE means:")
print("There is no fixed number of possible measurements.")
```
