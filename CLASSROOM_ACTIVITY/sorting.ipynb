```python
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------
# 1. Create DataFrames
# ----------------------------

students = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Name": ["Arun", "Bala", "Cathy", "David", "Eva"],
    "Department": ["CSE", "ECE", "CSE", "ECE", "CSE"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Marks": [85, 65, 90, 70, 78]
})

# ----------------------------
# 2. Combining
# ----------------------------

df = pd.merge(students, marks, on="ID")

print("Combined Data")
print(df)

# ----------------------------
# 3. Indexing
# ----------------------------

print("\nFirst Student")
print(df.iloc[0])

print("\nName and Marks")
print(df[["Name", "Marks"]])

# ----------------------------
# 4. Filtering
# ----------------------------

print("\nStudents with Marks > 75")
print(df[df["Marks"] > 75])

# ----------------------------
# 5. Grouping
# ----------------------------

print("\nDepartment Average")
print(df.groupby("Department")["Marks"].mean())

# ----------------------------
# 6. Sorting
# ----------------------------

sorted_df = df.sort_values(
    "Marks",
    ascending=False
)

print("\nSorted Marks")
print(sorted_df)

# ----------------------------
# 7. File I/O
# ----------------------------

df.to_csv(
    "student_analysis.csv",
    index=False
)

print("\nFile Saved Successfully")

# ----------------------------
# 8. Plotting
# ----------------------------

df.plot(
    x="Name",
    y="Marks",
    kind="bar"
)

plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.show()
```

```python

df.sort_values("Marks")
Default:
Low → High
Descending:
df.sort_values("Marks", ascending=False)
Result:
High → Low
Sort names alphabetically:
df.sort_values("Name")
Sort using multiple columns:
df.sort_values(["City", "Marks"])

```
