# Experiment 3: ID3 Decision Tree Algorithm

**Aim:** Demonstrate the working of the decision tree based ID3 algorithm. Build a decision tree using an appropriate dataset and use it to classify a new sample.

```python
# Step 1: Create the PlayTennis dataset
import pandas as pd

data = {
    'Outlook':    ['Sunny','Sunny','Overcast','Rain','Rain','Rain','Overcast',
                    'Sunny','Sunny','Rain','Sunny','Overcast','Overcast','Rain'],
    'Temperature':['Hot','Hot','Hot','Mild','Cool','Cool','Cool',
                    'Mild','Cool','Mild','Mild','Mild','Hot','Mild'],
    'Humidity':   ['High','High','High','High','Normal','Normal','Normal',
                    'High','Normal','Normal','Normal','High','Normal','High'],
    'Wind':       ['Weak','Strong','Weak','Weak','Weak','Strong','Strong',
                    'Weak','Weak','Weak','Strong','Strong','Weak','Strong'],
    'PlayTennis': ['No','No','Yes','Yes','Yes','No','Yes',
                    'No','Yes','Yes','Yes','Yes','Yes','No']
}
df = pd.DataFrame(data)
df.to_csv('play_tennis.csv', index=False)
df
```

```python
# Step 2: Encode categorical data and build the ID3 tree using sklearn
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier, export_text
import pandas as pd

df = pd.read_csv('play_tennis.csv')
encoders = {}
df_enc = df.copy()
for col in df.columns:
    le = LabelEncoder()
    df_enc[col] = le.fit_transform(df[col])
    encoders[col] = le

X = df_enc.drop('PlayTennis', axis=1)
y = df_enc['PlayTennis']

model = DecisionTreeClassifier(criterion='entropy', random_state=0)  # entropy => ID3 style
model.fit(X, y)

print(export_text(model, feature_names=list(X.columns)))
```

```python
# Step 3: Classify a new sample
new_sample = {'Outlook': 'Sunny', 'Temperature': 'Cool', 'Humidity': 'High', 'Wind': 'Strong'}
encoded_sample = [encoders[col].transform([new_sample[col]])[0] for col in X.columns]

prediction = model.predict([encoded_sample])[0]
result = encoders['PlayTennis'].inverse_transform([prediction])[0]

print("New sample:", new_sample)
print("Predicted class (PlayTennis):", result)
```
