# Experiment 2: Candidate-Elimination Algorithm

**Aim:** For a given set of training data examples stored in a .CSV file, implement and demonstrate the Candidate-Elimination algorithm to output the version space (set of all hypotheses consistent with the training examples).

```python
# Step 1: Create and save the training dataset
import pandas as pd

data = {
    'Sky':     ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity':['Normal', 'High', 'High', 'High'],
    'Wind':    ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water':   ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast':['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}
df = pd.DataFrame(data)
df.to_csv('training_data.csv', index=False)
df
```

```python
# Step 2: Implement the Candidate-Elimination algorithm
import numpy as np
import pandas as pd

def candidate_elimination(csv_path):
    df = pd.read_csv(csv_path)
    concepts = df.iloc[:, :-1].values
    target = df.iloc[:, -1].values
    n_attr = concepts.shape[1]

    specific_h = concepts[0].copy()
    general_h = [['?' for _ in range(n_attr)] for _ in range(n_attr)]

    for i, instance in enumerate(concepts):
        if target[i] == 'Yes':
            for x in range(n_attr):
                if instance[x] != specific_h[x]:
                    specific_h[x] = '?'
                    general_h[x][x] = '?'
        else:
            for x in range(n_attr):
                if instance[x] != specific_h[x]:
                    general_h[x][x] = specific_h[x]
                else:
                    general_h[x][x] = '?'

    # Remove overly general hypotheses (all '?')
    general_h = [h for h in general_h if h != ['?'] * n_attr]
    return specific_h, general_h

S, G = candidate_elimination('training_data.csv')
print("Final Specific Hypothesis (S):", list(S))
print("\nFinal General Hypotheses (G):")
for h in G:
    print(" ", h)
```
