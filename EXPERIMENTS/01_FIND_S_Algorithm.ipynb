# Experiment 1: FIND-S Algorithm

**Aim:** Implement and demonstrate the FIND-S algorithm to find the most specific hypothesis consistent with the given training examples.

```python
# Step 1: Create the training dataset (Enjoy Sport example) and save as CSV
import pandas as pd

data = {
    'Sky':     ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity':['Normal', 'High', 'High', 'High'],
    'Wind':    ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water':   ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast':['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}
df = pd.DataFrame(data)
df.to_csv('training_data.csv', index=False)
df
```

```python
# Step 2: Implement the FIND-S algorithm
import numpy as np

def find_s(csv_path):
    df = pd.read_csv(csv_path)
    attributes = df.columns[:-1]
    target = df.columns[-1]

    # Initialize hypothesis to the most specific one
    hypothesis = ['0'] * len(attributes)

    for i, row in df.iterrows():
        if row[target] == 'Yes':
            for j, attr in enumerate(attributes):
                if hypothesis[j] == '0':
                    hypothesis[j] = row[attr]
                elif hypothesis[j] != row[attr]:
                    hypothesis[j] = '?'
    return hypothesis, list(attributes)

final_hypothesis, attributes = find_s('training_data.csv')
print("Attributes      :", attributes)
print("Final Hypothesis:", final_hypothesis)
```

```python
# Step 3: Display result neatly
result = dict(zip(attributes, final_hypothesis))
print("Most Specific Hypothesis consistent with the positive examples:\n")
for k, v in result.items():
    print(f"  {k:10s} : {v}")
```
