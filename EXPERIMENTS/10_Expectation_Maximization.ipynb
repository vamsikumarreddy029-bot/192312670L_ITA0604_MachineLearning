# Experiment 10: Expectation-Maximization (EM) Algorithm

**Aim:** Write a Python program to implement the Expectation-Maximization algorithm for clustering data using a Gaussian Mixture Model.

```python
# Step 1: Create a synthetic dataset with two clusters
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
cluster1 = np.random.normal(loc=[2, 2], scale=0.6, size=(100, 2))
cluster2 = np.random.normal(loc=[7, 7], scale=0.8, size=(100, 2))
X = np.vstack([cluster1, cluster2])

plt.scatter(X[:, 0], X[:, 1], s=15, color='gray')
plt.title('Unlabeled Data')
plt.show()
```

```python
# Step 2: Apply the EM algorithm via Gaussian Mixture Model
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(n_components=2, random_state=42)
gmm.fit(X)
labels = gmm.predict(X)

print("Converged:", gmm.converged_)
print("Number of iterations:", gmm.n_iter_)
print("Means of the two Gaussians:\n", gmm.means_)
```

```python
# Step 3: Visualize the clustering result
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=15)
plt.scatter(gmm.means_[:, 0], gmm.means_[:, 1], c='red', marker='X', s=200, label='Cluster centers')
plt.title('EM (Gaussian Mixture Model) Clustering')
plt.legend()
plt.show()
```
