# Experiment 9: Compare Linear and Polynomial Regression

**Aim:** Compare the performance of Linear Regression and Polynomial Regression using Python.

```python
# Step 1: Create a synthetic non-linear dataset
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
X = np.sort(6 * np.random.rand(100, 1) - 3, axis=0)
y = 0.5 * X[:, 0]**2 + X[:, 0] + 2 + np.random.randn(100)

plt.scatter(X, y, color='gray')
plt.title('Synthetic Dataset')
plt.xlabel('X'); plt.ylabel('y')
plt.show()
```

```python
# Step 2: Fit Linear Regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

lin_reg = LinearRegression()
lin_reg.fit(X, y)
y_pred_lin = lin_reg.predict(X)

print("Linear Regression  -> MSE:", mean_squared_error(y, y_pred_lin),
      " R2:", r2_score(y, y_pred_lin))
```

```python
# Step 3: Fit Polynomial Regression (degree 2)
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

poly_reg = make_pipeline(PolynomialFeatures(degree=2), LinearRegression())
poly_reg.fit(X, y)
y_pred_poly = poly_reg.predict(X)

print("Polynomial Regression -> MSE:", mean_squared_error(y, y_pred_poly),
      " R2:", r2_score(y, y_pred_poly))
```

```python
# Step 4: Compare both fits visually
plt.scatter(X, y, color='gray', label='Data')
plt.plot(X, y_pred_lin, color='red', label='Linear Fit')
plt.plot(X, y_pred_poly, color='green', label='Polynomial Fit (deg=2)')
plt.legend()
plt.title('Linear vs Polynomial Regression')
plt.xlabel('X'); plt.ylabel('y')
plt.show()
```
